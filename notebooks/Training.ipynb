## Deep Learning Training

```python

train_keys=['Train_1', 'Train_2','Train_3','Train_4','Train_5','Train_6']
O1={}

for d_idx in range(7):
	d=0.1*d_idx
	dname=f"output_d_{d:.1f}"

	for GT in ['GT1','GT2','GT12']:
			Ydata=[]
			for z in [0.5]:
					for z0 in [0,1]:
							for k in ['MeanX','LocalX','GlobalX','NoSmooth']:
									y="Y_" + GT + "_x" + str(z0) + "_z" + str(z) + "_"
									Ydata.append(y+k)
			Xdata=['dataX','dataXNorm']
			spots= GT + '_overlapSpots'
			modelsList=['Autoencoder_dropout','Encoder_dropout']
			dg=GT + '_' + dname


			O1[dg]=train.BacterialClassifierWorkflow(spdata_i=spdata,train_keys=train_keys,Xdata=Xdata,Ydata=Ydata,spots=spots,signal_cutoff=[0.7],models=modelsList,n_epochsAE=200,kclassloss=10,n_epochs=50,dropout=d)
			gc.collect()
			torch.cuda.empty_cache()
			print(torch.cuda.memory_allocated() / 1024**2, "MB allocated")
			print(torch.cuda.memory_reserved() / 1024**2, "MB reserved")

dump(O1, 'models/CLASS_AEDmENCd' + dname +'_TRAIN123456.h5ad')
```

## Machine Learning

```python

train_keys=['Train_1', 'Train_2','Train_3','Train_4','Train_5','Train_6']
O1={}

d=0.0
dname='0.0'

for GT in ['GT1','GT2','GT12']:
		Ydata=[]
		for z in [0.5]:
				for z0 in [0,1]:
						for k in ['MeanX','LocalX','GlobalX','NoSmooth']:
								y="Y_" + GT + "_x" + str(z0) + "_z" + str(z) + "_"
								Ydata.append(y+k)
		Xdata=['dataX','dataXNorm']
		spots= GT + '_overlapSpots'
		modelsList=['RandomForest','Logistic','XGBoost']
		dg=GT + '_' + dname


		O1[dg]=train.BacterialClassifierWorkflow(spdata_i=spdata,train_keys=train_keys,Xdata=Xdata,Ydata=Ydata,spots=spots,signal_cutoff=[0.7],models=modelsList,n_epochsAE=200,kclassloss=10,n_epochs=50,dropout=d)
		gc.collect()
dump(O1, 'models/CLASS_LOGmRFmXGBd' + dname +'_TRAIN123456.h5ad')
```
